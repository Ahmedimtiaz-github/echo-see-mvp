```python

import os
import librosa
import soundfile as sf
input_folder = "../demo_audio"
for file in os.listdir(input_folder):
    if file.endswith(".mp3") or file.endswith(".wav"):
        path = os.path.join(input_folder, file)

        # Load audio and resample to 16k
        audio, sr = librosa.load(path, sr=16000)

        # Output file name
        new_name = os.path.splitext(file)[0] + "_16k.wav"
        output_path = os.path.join(input_folder, new_name)

        # Save WAV file
        sf.write(output_path, audio, 16000)

        print(f"Converted: {file} -> {new_name}")
```

```python
import os

folder = "../demo_audio"

for file in os.listdir(folder):
    path = os.path.join(folder, file)

    # Delete ONLY original non-converted files
    if file.endswith(".mp3") or (file.endswith(".wav") and not file.endswith("_16k.wav")):
        print(f"Deleting original file: {file}")
        os.remove(path)

print("Cleanup completed! Only 16k WAV files remain.")
```
